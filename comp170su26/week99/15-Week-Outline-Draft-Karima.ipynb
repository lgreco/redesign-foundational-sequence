# COMP 170 Fall 2026 — Draft Weekly Outline

## Notes on this draft

This draft builds on the original Fall outline and preserves its main examples and overall direction.

- Functions, parameters, and return values are introduced in Week 3 and reinforced in later examples.
- Function design and contracts are developed more explicitly in Week 8.
- Relational expressions are made explicit in Week 6 and revisited in later applications.
- A light introduction to object-oriented programming is added in Week 9 through a bank-account example and reused in the ATM application.
- The outline is adapted to the Fall 2026 MWF holiday calendar.
- A few secondary topics are marked optional to keep shortened weeks manageable.
